```python
import requests
import json
import pandas as pd

# Test RapidAPI Cricbuzz endpoint with fallback test payload
sample_payload = {
    "match_id": 999,
    "series": "India vs Australia T20I Series 2026",
    "venue": "Narendra Modi Stadium, Ahmedabad",
    "batters": [
        {"name": "Virat Kohli", "runs": 76, "balls": 48},
        {"name": "Rohit Sharma", "runs": 64, "balls": 38}
    ]
}

print("✅ Ingestion payload verified:")
print(json.dumps(sample_payload, indent=2))
```

```python
df_batters = pd.DataFrame(sample_payload["batters"])
df_batters["strike_rate"] = (df_batters["runs"] / df_batters["balls"]) * 100
df_batters.round(2)
```
